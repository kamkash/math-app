# Refactoring Basic Struct in Rust
This notebook demonstrates how to refactor the `lib.rs` file by moving the `Basic` struct and its implementations to a separate `basic.rs` module.

## Setup Project Structure
Create a Rust project with a `src` directory and ensure `lib.rs` exists. Add a new file named `basic.rs` in the `src` directory.

```python
# Create the Rust project structure
!cargo new math_app --lib
!touch math_app/src/basic.rs
```

## Define Basic Struct in basic.rs
Move the definition of the `Basic` struct from `lib.rs` to `basic.rs`.

```python
# Define the Basic struct in basic.rs
%%writefile math_app/src/basic.rs
pub struct Basic {
    pub value: i32,
}

impl Basic {
    pub fn new(value: i32) -> Self {
        Basic { value }
    }
}
```

## Implement Methods for Basic Struct
Move all implementations of the `Basic` struct from `lib.rs` to `basic.rs`.

```python
# Add additional methods for Basic struct in basic.rs
%%writefile -a math_app/src/basic.rs

impl Basic {
    pub fn increment(&mut self) {
        self.value += 1;
    }

    pub fn display(&self) {
        println!("Value: {}", self.value);
    }
}
```

## Refactor lib.rs to Use basic.rs
Update `lib.rs` to include the `basic` module and re-export the `Basic` struct if necessary.

```python
# Refactor lib.rs to use the basic module
%%writefile math_app/src/lib.rs
pub mod basic;

pub use basic::Basic;
```

## Test the Refactored Code
Write or run existing tests to ensure the refactored code works as expected.

```python
# Write a test to verify the refactored code
%%writefile math_app/tests/test_basic.rs
use math_app::Basic;

#[test]
fn test_basic_struct() {
    let mut basic = Basic::new(10);
    assert_eq!(basic.value, 10);
    basic.increment();
    assert_eq!(basic.value, 11);
}
```

```python
# Run the tests
!cargo test
```
